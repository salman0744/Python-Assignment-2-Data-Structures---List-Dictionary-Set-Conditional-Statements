```python

# 1. List Creation
# a
age_list = [24, 25, 26, 27, 28]  

# b
name_list = ["Alice", "Bob", "Charlie", "David", "Eva"]  

# 2. List Operations/Modifications
# a
name_list.append("Yazhini")  

# b
age_list.insert(2, 30)  

# c
name_list.remove("Yazhini")  

# d
age_list.pop()  

# e
age_list.extend([29, 30, 26])  

# f
age_list.sort(reverse=True)  

# g
max_age = max(age_list)  
min_age = min(age_list)  
sum_age = sum(age_list)  

print("Updated age_list:", age_list)
print("Max Age:", max_age, "| Min Age:", min_age, "| Sum of Ages:", sum_age)

# 3. Accessing List Elements
# a
print("First element:", name_list[0])  

# b
print("Last element:", name_list[-1])  

# c
print("Elements index 2 to 4:", name_list[2:5])  

# d
print("Reversed name_list:", name_list[::-1])  
```

```python
#Dictionary (Creation, Modification and Access):
# a
student_marks = {
    "Alice": 85,
    "Bob": 72,
    "Charlie": 90,
    "David": 68,
    "Eva": 95
}  

# b
print("Bob's mark:", student_marks["Bob"])  

# c
student_marks["Janani"] = 80  

# d
student_marks["Alice"] = 82  

# e
print("Keys:", list(student_marks.keys()))  
print("Values:", list(student_marks.values()))  
print("Key-Value Pairs:", list(student_marks.items()))  
```

```python
#SETS operations:
# a
my_set = set(['a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'])  
print("my_set:", my_set)  

# b
try:
    my_set[4] = 's'  
except TypeError as e:
    print("Error:", e)  

# c
set1 = {1, 3, 5, 7, 9}  
set2 = {2, 3, 5, 8, 10}  

# d
union_set = set1.union(set2)  
intersection_set = set1.intersection(set2)  

print("Union of set1 and set2:", union_set)  
print("Intersection of set1 and set2:", intersection_set) 
```

```python
#Operators & Conditional Statements :
#(IF, ELIF, ELSE)
score = float(input("Enter your score (0 to 10): "))  

if score > 7:  #
    print("Above Average: Excellent work! Outstanding performance.")  #
elif 4 <= score <= 7:  #
    print("Average: Good effort! Keep practicing, there's room for improvement.")  #
elif score < 4:  #
    print("Below Average: Need to improve your performance, consistent practice will lead to better results.")  #
else:
    print("Invalid score entered.")
```
